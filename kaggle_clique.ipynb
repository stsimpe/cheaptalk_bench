# ΒΗΜΑΤΑ 2–6 — Clique — το τρίτο σημείο του RQ2

**Τι κλείνει.** Το RQ2 έχει σήμερα δύο σημεία, αστέρα και δακτύλιο, και ένα
ασθενές αποτέλεσμα (p ≈ 0,19). Ο μηχανισμός που εξηγούσε το «γιατί» — ότι ο
κόμβος-κέντρο είναι σημείο κατάρρευσης — **διαψεύστηκε**: η διαφορά κέντρου
μείον φύλλων είναι −0,00 έως +0,05 και τα δύο κινούνται μαζί με r = 0,97.

Αυτό που επιβίωσε είναι η **αραίωση της εμπειρίας**: ένα φύλλο του αστέρα
παίζει μία ακμή, οπότε ένας αποστάτης είναι το 100 % όσων βλέπει· στον
δακτύλιο παίζει δύο και ο ίδιος αποστάτης είναι το 50 %. Η clique είναι η
ευθεία δοκιμασία αυτής της υπόθεσης, γιατί εκεί κάθε πράκτορας παίζει **τρεις**
ακμές και ο αποστάτης πέφτει στο 33 %.

**Η πρόβλεψη, διατυπωμένη πριν τρέξει οτιδήποτε:** αν η αραίωση είναι ο
μηχανισμός, στα κελιά χαμηλής επικοινωνίας η συνεργασία πρέπει να αυξάνεται
μονότονα, **αστέρας < δακτύλιος < clique**.

| μοντέλο | αστέρας | δακτύλιος | πρόβλεψη για clique |
|---|---|---|---|
| gemma-2-9b | 0,08 | 0,28 | > 0,28 |
| Llama-3.1-8B | 0,47 | 0,57 | > 0,57 |
| Qwen2.5-7B | 0,08 | 0,11 | > 0,11 |
| Qwen3-4B | 0,04 | 0,05 | > 0,05 |
| **gemma-2-2b** | **0,67** | **0,18** | **> 0,18 — το κρίσιμο** |

Το gemma-2-2b είναι η μοναδική αντιστροφή ολόκληρης της καμπάνιας. Αν η
clique του βγει κοντά στο 0,18, η αραίωση κερδίζει και ο αστέρας του ήταν το
περίεργο. Αν βγει κοντά στο 0,67, ο βαθμός του κόμβου δεν είναι ο άξονας και
χρειαζόμαστε άλλη εξήγηση. **Και οι δύο εκβάσεις είναι δημοσιεύσιμες.**

**Δεύτερος λόγος.** Η clique είναι η φυσική δομή ενός καρτέλ: όλοι μιλούν με
όλους. Είναι η τοπολογία που θα χρειαστεί το παιχνίδι τιμολόγησης Bertrand
στο επόμενο στάδιο, οπότε τη μετράμε πρώτα εδώ, σε παιχνίδι που ήδη ξέρουμε.

## Ο έλεγχος που κρίνει το πείραμα

Χτίζει το prompt μιας clique και επιβεβαιώνει ότι περιγράφει clique, ότι δεν
έχει μείνει κείμενο αστέρα, και ότι ο πράκτορας έχει όντως τρεις γείτονες.
Κοστίζει μηδέν.


```python
import sys
sys.path.insert(0, '.')
from games import GAMES
from topology import make_topology
from prompts import build_system_prompt

topo = make_topology('clique', 4)
assert topo.neighbors(0) == [1, 2, 3], topo.neighbors(0)

common = dict(game=GAMES['pd'], condition='cheap_talk', n_neighbors=3,
              total_agents=4, topology_text=topo.describe(0))
old = build_system_prompt(**common)
new = build_system_prompt(**common, communication_text=topo.describe_communication(0))

assert 'central agent' in old, 'το παλιό prompt δεν έχει το κείμενο του αστέρα;'
assert 'central agent' not in new and 'peripheral' not in new, \
    'το prompt της clique ΕΧΕΙ ακόμα κείμενο αστέρα -- σταμάτα'
assert 'fully connected' in new and 'reaches every other agent' in new, \
    'το prompt δεν περιγράφει clique -- σταμάτα'

diff = [l for l in new.splitlines() if l not in old.splitlines()]
print('η ΜΟΝΗ γραμμή που αλλάζει σε σχέση με το παλιό prompt:\n')
for l in diff:
    print(' ', l.strip()[:200])
assert len(diff) == 1, 'άλλαξαν περισσότερα από μία παράγραφο -- σταμάτα'
```

## Τι θα τρέξει

Ακριβώς τα ίδια κελιά με το ablation του δακτυλίου, ώστε οι τρεις τοπολογίες
να συγκρίνονται γραμμή προς γραμμή: **clique, PD μόνο, n=5**, σενάρια
`silence`, `no_sense`, `counterfactual`, `baseline`. **25 runs, 2.240 κλήσεις
ανά μοντέλο.**

Τα τρία πρώτα είναι τα κελιά χαμηλής επικοινωνίας όπου ζει η πρόβλεψη της
αραίωσης· το `baseline` δίνει το ταβάνι με ανοιχτό κανάλι και, μέσω του
`no_comm` σκέλους του, την άγκυρα κάθε μοντέλου στη νέα τοπολογία.

**Προτεινόμενη σειρά:** ξεκίνα από τα δύο άκρα του άξονα, gemma-2-2b (η
μοναδική αντιστροφή, και το φθηνότερο) και gemma-2-9b (το μεγαλύτερο θετικό
άνοιγμα). Αν η μονοτονία φανεί εκεί, συνέχισε με τα υπόλοιπα τρία.


```python
# ΒΗΜΑΤΑ 2-6 -- ξεσχολίασε ΕΝΑ μπλοκ κάθε φορά, με αυτή τη σειρά.
# Το gemma-2-2b ΔΕΝ είναι εδώ: η clique του τρέχει στο βήμα 1
# (kaggle_g2b_triad.ipynb), μαζί με αστέρα και δακτύλιο στην ίδια συνεδρία.
#
# Η clique είναι ~45% πιο αργή από τον αστέρα και το Kaggle κόβει στις 12
# ώρες· κανένα μπλοκ δεν ξεπερνά τις ~10.

# --- βήμα 2 -------------------------------------------------------------
MODELS    = ['Qwen/Qwen2.5-7B-Instruct']                           # ~4,5 h
SCENARIOS = ['silence', 'no_sense', 'counterfactual', 'baseline']

# --- βήμα 3 -------------------------------------------------------------
# MODELS    = ['meta-llama/Llama-3.1-8B-Instruct']                 # ~8,5 h
# SCENARIOS = ['silence', 'no_sense', 'counterfactual', 'baseline']

# --- βήμα 4 -------------------------------------------------------------
# MODELS    = ['Qwen/Qwen3-4B']                                    # ~9,9 h
# SCENARIOS = ['silence', 'no_sense', 'counterfactual', 'baseline']

# --- βήματα 5 και 6: το gemma-2-9b δεν χωράει ολόκληρο (~11,8 h) -------
# Δύο κομμάτια, ίδια κελιά, ΙΔΙΟΣ φάκελος· το campaign.py κρίνει μόνο τα
# σενάρια που του ζητήθηκαν, οπότε το βήμα 6 δεν μπερδεύεται από το 5.
# MODELS    = ['google/gemma-2-9b-it']                             # ~6,7 h
# SCENARIOS = ['silence', 'no_sense', 'counterfactual']
#
# MODELS    = ['google/gemma-2-9b-it']                             # ~5,1 h
# SCENARIOS = ['baseline']

SESSION   = 'A'
TOPOLOGY  = 'clique'
GAMES     = ['pd']

WALL_HOURS = 10.5

def args_for(model):
    return ['--model', model, '--session', SESSION, '--topology', TOPOLOGY,
            '--scenarios', *SCENARIOS, '--games', *GAMES,
            '--topology-aware-comm-prompt']

expected = {'silence': 5, 'no_sense': 5, 'counterfactual': 5, 'baseline': 10}
print(f'{len(MODELS)} μοντέλα, {TOPOLOGY}, {GAMES}')
print('σενάρια:', SCENARIOS, '->', sum(expected[s] for s in SCENARIOS), 'runs ανά μοντέλο')
for m in MODELS:
    print('  ', m)
```

## Preview — κοστίζει μηδέν

Σε κάθε μοντέλο πρέπει να δεις `topology : clique`, τη γραμμή `PROMPT`,
`expecting : 25 run files` και φάκελο που τελειώνει σε `_clique_commfix`.


```python
import subprocess, sys

ok = True
for m in MODELS:
    print('=' * 70)
    r = subprocess.run([sys.executable, 'campaign.py', *args_for(m), '--dry-run'],
                       capture_output=True, text=True)
    out = r.stdout.strip()
    print(out[:1000])
    if r.returncode != 0:
        ok = False
        print('ΣΦΑΛΜΑ:', r.stderr.strip()[-300:])
    want_runs = sum(expected[s] for s in SCENARIOS)
    for needle in ('topology-aware', 'clique', f'expecting    : {want_runs}'):
        if needle not in out:
            ok = False
            print(f'ΛΕΙΠΕΙ ΑΠΟ ΤΟ ΠΛΑΝΟ: {needle!r}')

assert ok, 'Κάποιο plan απέτυχε -- μη συνεχίσεις.'
print('\n' + '=' * 70)
print('όλα τα πλάνα εντάξει')
```

## Εκτέλεση


```python
import subprocess, sys, time

t0 = time.time()
results = []

for m in MODELS:
    elapsed_h = (time.time() - t0) / 3600
    if elapsed_h > WALL_HOURS:
        print(f'\n[STOP] {elapsed_h:.1f} h -- δεν ξεκινάω το {m}, δεν προλαβαίνει.')
        results.append((m, 'skipped', 0.0))
        continue

    print('\n' + '=' * 70)
    print(f'{m}   ({elapsed_h:.1f} h μέχρι τώρα)')
    print('=' * 70, flush=True)

    t = time.time()
    r = subprocess.run([sys.executable, 'campaign.py', *args_for(m)])
    mins = (time.time() - t) / 60
    status = 'OK' if r.returncode == 0 else f'FAILED (exit {r.returncode})'
    results.append((m, status, mins))
    print(f'\n--> {m}: {status}, {mins:.0f} λεπτά', flush=True)

print('\n' + '=' * 70)
print('ΑΠΟΛΟΓΙΣΜΟΣ')
for m, s, mins in results:
    print(f'  {s:22s} {mins:6.0f} λ   {m}')
print(f'\nσύνολο {(time.time() - t0) / 3600:.1f} h')
```

## Έλεγχος των δεδομένων και πρώτη ματιά

Επιβεβαιώνει ότι γράφτηκε clique και σωστό prompt, και τυπώνει αμέσως το
ποσοστό συνεργασίας ανά κελί — ώστε να δεις την πρόβλεψη πριν καν κατεβάσεις
τα αρχεία.


```python
import glob, json, sys, collections, statistics
sys.path.insert(0, '.')
from analysis import summarise_run, scenario_of

rows = collections.defaultdict(list)
for d in sorted(glob.glob('/kaggle/working/results/*_clique_commfix')):
    for p in glob.glob(f'{d}/**/*.json', recursive=True):
        rec = json.load(open(p, encoding='utf-8'))
        cfg = rec['config']
        if cfg['game'] != 'pd':
            continue
        assert rec['topology']['type'] == 'clique', p
        assert cfg.get('topology_aware_comm_prompt') is True, f'ΛΑΘΟΣ PROMPT: {p}'
        model = cfg['model']['model_id'].split('/')[-1]
        cell = scenario_of(rec)
        rows[(model, cell)].append(summarise_run(rec)['coop_rate_overall'])

print(f"{'model':24s} {'cell':22s} {'n':>3s} {'coop':>6s}")
for k in sorted(rows):
    v = rows[k]
    print(f'{k[0]:24s} {k[1]:22s} {len(v):3d} {statistics.fmean(v):6.3f}')
```
