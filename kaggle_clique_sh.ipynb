# Stag Hunt στην clique — η τρίτη τοπολογία του δεύτερου παιγνίου

**Τι κλείνει.** Το PD έχει τρεις τοπολογίες, το Stag Hunt δύο: αστέρα και
δακτύλιο. Και ο δακτύλιος του Stag Hunt κουβαλά ακόμα την παλιά πρόταση
δρομολόγησης (το prompt bug της 20/9) σε όλα τα κελιά με ανοιχτό κανάλι. Η
clique με το διορθωμένο prompt δίνει στο Stag Hunt **δεύτερη καθαρή τοπολογία**
δίπλα στον αστέρα και κλείνει τον πίνακα 3 τοπολογίες × 2 παίγνια. Ίδια 14
κελιά με την clique του PD, 5 μοντέλα, n=5: **350 runs**.

**Τι ξέρουμε ήδη, χωρίς GPU (7/10).**

- Το ισοζύγιο του Stag Hunt **μόνο στον αστέρα**, που δεν έχει το bug, δίνει
  την ίδια εικόνα με το συνολικό: ουσιαστικό περιεχόμενο +0,03, επιθετικό
  περιεχόμενο στα μηνύματα −0,16 (βλαπτικό σε 3/5), 12 από τα 13 βλαπτικά
  κελιά στα δύο Gemma. Ο δακτύλιος μόνος του δίνει το ίδιο (+0,03, −0,17,
  12/14). Τα συμπεράσματα του Stag Hunt **δεν** στηρίζονται στο λάθος prompt.
- Στα κελιά χαμηλής επικοινωνίας αστέρας και δακτύλιος δεν διαφέρουν: ο
  δακτύλιος είναι ψηλότερα σε 6 και χαμηλότερα σε 8 από τα 15.
- Τρία μοντέλα είναι στο ταβάνι σχεδόν παντού: Qwen2.5 και Qwen3-4B στο
  0,93–1,00, το gemma-2-9b στα περισσότερα κελιά. Πληροφορία υπάρχει κυρίως
  στο Llama και στο gemma-2-2b.

**Άρα τι περιμένουμε.** Επιβεβαίωση, όχι νέο εύρημα: ότι το ισοζύγιο του Stag
Hunt επαναλαμβάνεται σε τρίτη τοπολογία με σωστό prompt, και ότι η τοπολογία
δεν κινεί το Stag Hunt. Οι κανόνες ανάγνωσης γράφτηκαν πριν τρέξει οτιδήποτε,
στην επόμενη ενότητα.

## Κανόνες ανάγνωσης (7 Οκτωβρίου 2026, πριν τρέξει οτιδήποτε)

Μονάδα ο μέσος όρος run ανά (μοντέλο, κελί), Stag Hunt, clique. Δ = κελί −
`no_comm` της clique του ίδιου μοντέλου, ζώνη ±0,05, κατηγορίες περιεχομένου
όπως στο `harm_ledger.py`.

1. **Επανάληψη του ισοζυγίου** (κρίνεται μόνο όταν υπάρχουν και τα τρία
   sessions). Επαναλαμβάνεται αν ισχύουν **και τα τρία**:
   - (α) το επιθετικό περιεχόμενο στα μηνύματα (`framing_competitive`) έχει τον
     πιο αρνητικό μέσο Δ από τις τέσσερις κατηγορίες·
   - (β) το ουσιαστικό περιεχόμενο έχει μέσο Δ μέσα στο ±0,05·
   - (γ) τουλάχιστον τα 2/3 των βλαπτικών κελιών είναι στα δύο Gemma.

   Αστέρας και δακτύλιος περνούν και τα τρία. Τα (α)–(γ) υπολογίζονται όπως
   εκεί, με όλα τα κελιά, και ξανά χωρίς τα κελιά με > 15% άκυρες αποφάσεις·
   το συμπέρασμα γράφεται μόνο αν ισχύει και στις δύο εκδοχές. Αν αποτύχει
   κάποιο, γράφεται ποιο, και η διατριβή λέει ότι το ισοζύγιο του Stag Hunt
   ισχύει σε δύο από τις τρεις τοπολογίες.
2. **Τοπολογία** (κρίνεται μετά το session A). Στα 15 κελιά χαμηλής
   επικοινωνίας (`no_comm`, `silence`, `no_sense` × 5 μοντέλα), sign test
   clique έναντι δακτυλίου και clique έναντι αστέρα. Επίδραση τοπολογίας
   δηλώνεται μόνο αν **και οι δύο** συγκρίσεις δίνουν τουλάχιστον 12 από τα 15
   κελιά προς την ίδια κατεύθυνση (οι ισοπαλίες μετρούν κατά). Αλλιώς: «η
   τοπολογία δεν κινεί το Stag Hunt», όπως ήδη ο αστέρας έναντι του
   δακτυλίου.
3. **Άκυρες αποφάσεις.** Κελί με > 15% άκυρες αναφέρεται με επιφύλαξη. Το
   gemma-2-2b είχε 45% και 65% στο `silence` του Stag Hunt σε αστέρα και
   δακτύλιο, και ~21% στο `no_sense`· αναμένεται το ίδιο.
4. **Τι δεν κρίνεται εδώ.** Η θέση του frame (RQ3) στο Stag Hunt δεν είναι
   ισχυρισμός της διατριβής. Τα κελιά του session C μπαίνουν στο ισοζύγιο
   χωρίς δικό τους κανόνα.

## Ο έλεγχος που κρίνει το πείραμα

Χτίζει το prompt της clique για το Stag Hunt, με ανοιχτό και με κλειστό κανάλι,
και επιβεβαιώνει: ότι είναι Stag Hunt με τις αποδόσεις του αστέρα και του
δακτυλίου, ότι περιγράφει clique, ότι δεν έχει μείνει κείμενο αστέρα, και ότι
η μόνη γραμμή που αλλάζει σε σχέση με το παλιό prompt είναι η δρομολόγηση.
Κοστίζει μηδέν.

```python
import sys
sys.path.insert(0, '.')
from games import GAMES
from topology import make_topology
from prompts import build_system_prompt

topo = make_topology('clique', 4)
assert topo.neighbors(0) == [1, 2, 3], topo.neighbors(0)

common = dict(game=GAMES['sh'], n_neighbors=3, total_agents=4, topology_text=topo.describe(0))
old = build_system_prompt(condition='cheap_talk', **common)
new = build_system_prompt(condition='cheap_talk', **common,
                          communication_text=topo.describe_communication(0))
closed = build_system_prompt(condition='no_comm', **common)

for name, p in (('cheap_talk', new), ('no_comm', closed)):
    rows = [l.strip() for l in p.splitlines() if l.strip().startswith('|')]
    assert '| Stag      | 6, 6 | 1, 4 |' in rows and '| Hare      | 4, 1 | 2, 2 |' in rows, \
        f'{name}: οι αποδόσεις δεν είναι του Stag Hunt του πλέγματος -- σταμάτα'
    assert 'Cooperate' not in p and 'Defect' not in p, f'{name}: ετικέτες PD στο prompt -- σταμάτα'
    assert 'fully connected' in p, f'{name}: το prompt δεν περιγράφει clique -- σταμάτα'
    assert 'central agent' not in p and 'peripheral' not in p, \
        f'{name}: το prompt ΕΧΕΙ ακόμα κείμενο αστέρα -- σταμάτα'
assert 'reaches every other agent' in new, 'η δρομολόγηση δεν είναι της clique -- σταμάτα'

diff = [l for l in new.splitlines() if l not in old.splitlines()]
print('η ΜΟΝΗ γραμμή που αλλάζει σε σχέση με το παλιό prompt:\n')
for l in diff:
    print(' ', l.strip()[:200])
assert len(diff) == 1, 'άλλαξαν περισσότερα από μία παράγραφο -- σταμάτα'
print('\nStag Hunt, clique, σωστή δρομολόγηση, και στα δύο σκέλη')
```

## Ποιο βήμα

```python
STEP = 1          # <-- ΜΟΝΟ ΑΥΤΟ ΑΛΛΑΖΕΙΣ (1-7, με αυτή τη σειρά)

G2B, Q25 = 'google/gemma-2-2b-it', 'Qwen/Qwen2.5-7B-Instruct'
LLA, Q3, G9B = 'meta-llama/Llama-3.1-8B-Instruct', 'Qwen/Qwen3-4B', 'google/gemma-2-9b-it'

# Το session C του gemma-2-9b (~10,6 h) δεν χωρά σε ένα βήμα: σπάει σε δύο.
C_BT = ['framing_business_context', 'framing_team_context']
C_COMP = ['framing_competitive_context']

# βήμα -> (κάρτα 0, κάρτα 1). Κάθε κάρτα τρέχει τις δουλειές της με τη σειρά.
# Δουλειά = (μοντέλο, session, σενάρια· None = όλο το session).
STEPS = {
    1: ([(G9B, 'A', None)], [(LLA, 'A', None), (Q25, 'A', None)]),
    2: ([(Q3,  'A', None)], [(G2B, 'A', None), (Q25, 'B', None)]),
    3: ([(G9B, 'B', None)], [(G2B, 'B', None)]),
    4: ([(Q3,  'B', None)], [(LLA, 'B', None)]),
    5: ([(G9B, 'C', C_BT)], [(LLA, 'C', None)]),
    6: ([(Q3,  'C', None)], [(G2B, 'C', None)]),
    7: ([(G9B, 'C', C_COMP)], [(Q25, 'C', None)]),
}
HOURS = {1: 8, 2: 8, 3: 7.5, 4: 6, 5: 7.5, 6: 8, 7: 5}

def args_for(model, session, scenarios):
    a = ['--model', model, '--session', session, '--topology', 'clique',
         '--games', 'sh', '--topology-aware-comm-prompt', '--zip-mirror', '']
    return a + ['--scenarios', *scenarios] if scenarios else a

def label(model, session, scenarios):
    part = ''
    if scenarios:
        part = ' (' + ', '.join(s.replace('framing_', '').replace('_context', '')
                                for s in scenarios) + ')'
    return f"{model.split('/')[-1]} · {session}{part}"

LANES = STEPS[STEP]
used = [m for lane in LANES for m, _, _ in lane]
# Κάθε μοντέλο μία φορά ανά βήμα: ο φάκελός του μετρά τα runs μιας δουλειάς, και
# τα αρχεία του στον δίσκο σβήνονται όταν τελειώσει.
assert len(used) == len(set(used)), 'ένα μοντέλο εμφανίζεται δύο φορές στο βήμα -- σταμάτα'
NEEDS_HF = any(m.startswith(('meta-llama/', 'google/')) for m in used)

from campaign import plan
JOBS, EXPECTED = [], {}
for lane in LANES:
    jobs = []
    for model, session, scenarios in lane:
        name = label(model, session, scenarios)
        EXPECTED[name] = plan(model, session, 'clique', 5, 16, '/kaggle/working/results',
                              only=scenarios, games=['sh'], comm_fix=True)['expected_runs']
        jobs.append((name, model, args_for(model, session, scenarios)))
    JOBS.append(jobs)

print(f'βήμα {STEP}, Stag Hunt, clique, ~{HOURS[STEP]} h')
for g, jobs in enumerate(JOBS):
    print(f'  κάρτα {g}: ' + '  ->  '.join(f'{n} [{EXPECTED[n]} runs]' for n, _, _ in jobs))
print(f'  σύνολο {sum(EXPECTED.values())} runs')
```

## Εκτέλεση: δύο κάρτες, μία ουρά η καθεμία

Κάθε κάρτα τρέχει τις δουλειές της με τη σειρά, η καθεμία ως ξεχωριστό
`campaign.py`. Η έξοδος κάθε δουλειάς πάει σε δικό της log στο
`/kaggle/working/logs/`. Εδώ τυπώνεται η γραμμή `device map` κάθε μοντέλου μόλις
φορτώσει, η πρόοδος κάθε 10 λεπτά, κάθε δουλειά που τελειώνει, και ο απολογισμός.

```python
import glob, os, shutil, subprocess, sys, threading, time

LOG_DIR, RESULTS = '/kaggle/working/logs', '/kaggle/working/results'
POLL_S, REPORT_S = 30, 600
os.makedirs(LOG_DIR, exist_ok=True)

try:
    from huggingface_hub import constants as _hfc
    HF_CACHE = _hfc.HF_HUB_CACHE
except Exception:
    HF_CACHE = os.path.expanduser('~/.cache/huggingface/hub')

def free_gb(path):
    while not os.path.exists(path):
        path = os.path.dirname(path)
    return shutil.disk_usage(path).free / 1e9

# Δύο μοντέλα στον δίσκο μαζί θέλουν ως ~35 GB (gemma-2-9b 18,5 + Llama 16).
# Κάτω από 45 GB τα αρχεία κάθε μοντέλου σβήνονται μόλις είναι στην κάρτα:
# μετά τη φόρτωση δεν τα ξαναδιαβάζει τίποτα.
FREE = free_gb(HF_CACHE)
DELETE_AFTER_LOAD = FREE < 45
print(f'ελεύθερος δίσκος {FREE:.0f} GB: τα αρχεία κάθε μοντέλου σβήνονται '
      f'{"μόλις φορτώσει" if DELETE_AFTER_LOAD else "όταν τελειώσει η δουλειά του"}')
assert FREE > 22, 'λιγότερα από 22 GB ελεύθερα: ούτε το gemma-2-9b δεν χωρά -- σταμάτα'

def cache_dir(model):
    return os.path.join(HF_CACHE, 'models--' + model.replace('/', '--'))

def runs_done(model):
    d = os.path.join(RESULTS, model.split('/')[-1] + '_clique_commfix')
    return sum(1 for p in glob.glob(f'{d}/**/*.json', recursive=True)
               if not os.path.basename(p).startswith('_') and f'{os.sep}zips{os.sep}' not in p)

def log_has(path, needle):
    try:
        with open(path, encoding='utf-8', errors='replace') as fh:
            return needle in fh.read()
    except FileNotFoundError:
        return False

load_lock = threading.Lock()   # ένα μοντέλο φορτώνει τη φορά
guard = threading.Lock()       # προστατεύει τα state / finished
state, finished = {}, []

def run_lane(gpu, jobs):
    env = dict(os.environ, CUDA_VISIBLE_DEVICES=str(gpu), PYTHONUNBUFFERED='1',
               TOKENIZERS_PARALLELISM='false', OMP_NUM_THREADS='2')
    for i, (name, model, extra) in enumerate(jobs):
        log = os.path.join(LOG_DIR, f"step{STEP}_gpu{gpu}_{i}_{model.split('/')[-1]}.log")
        t = time.time()
        try:
            with load_lock:
                with guard:
                    state[gpu] = dict(name=name, model=model, log=log, t0=t, phase='φόρτωση')
                with open(log, 'w', encoding='utf-8') as fh:
                    proc = subprocess.Popen([sys.executable, 'campaign.py', *extra],
                                            stdout=fh, stderr=subprocess.STDOUT, env=env)
                while proc.poll() is None and not log_has(log, '[local] Model loaded'):
                    time.sleep(10)
                if DELETE_AFTER_LOAD and proc.poll() is None:
                    shutil.rmtree(cache_dir(model), ignore_errors=True)
            with guard:
                state[gpu]['phase'] = 'runs'
            rc = proc.wait()
            status = 'OK' if rc == 0 else f'FAILED (exit {rc})'
        except Exception as e:   # ποτέ σιωπηλά: η δουλειά μπαίνει στον απολογισμό
            status = f'ERROR {e!r}'[:60]
        shutil.rmtree(cache_dir(model), ignore_errors=True)   # δεν ξαναχρειάζεται σε αυτό το βήμα
        with guard:
            finished.append((gpu, name, status, (time.time() - t) / 60, log))
    with guard:
        state[gpu] = None

threads = []
t0 = time.time()
for gpu, jobs in enumerate(JOBS):
    if jobs:
        th = threading.Thread(target=run_lane, args=(gpu, jobs), daemon=True)
        th.start()
        threads.append(th)
        time.sleep(2)   # η κάρτα 0 παίρνει πρώτη την κλειδαριά φόρτωσης

shown_maps, n_reported, last = set(), 0, time.time()
while any(th.is_alive() for th in threads):
    time.sleep(POLL_S)
    with guard:
        snap = {g: dict(s) for g, s in state.items() if s}
        done = list(finished)
    for g, s in snap.items():
        if s['log'] in shown_maps:
            continue
        try:
            with open(s['log'], encoding='utf-8', errors='replace') as fh:
                line = next((l for l in fh if '[local] device map' in l), None)
        except FileNotFoundError:
            line = None
        if line:
            shown_maps.add(s['log'])
            print(f"κάρτα {g} · {s['name']}: {line.strip()}", flush=True)
    for g, name, st, mins, _ in done[n_reported:]:
        print(f'--> κάρτα {g} · {name}: {st}, {mins:.0f} λεπτά', flush=True)
    n_reported = len(done)
    if time.time() - last >= REPORT_S:
        last = time.time()
        print(f'--- {(last - t0) / 3600:.1f} h ---', flush=True)
        for g, s in sorted(snap.items()):
            print(f"  κάρτα {g}: {s['name']:34s} {s['phase']:8s} "
                  f"{runs_done(s['model']):3d}/{EXPECTED[s['name']]} runs, "
                  f"{(time.time() - s['t0']) / 3600:.1f} h", flush=True)

for th in threads:
    th.join()
for g, name, st, mins, _ in finished[n_reported:]:
    print(f'--> κάρτα {g} · {name}: {st}, {mins:.0f} λεπτά')

print('\n' + '=' * 70)
print('ΑΠΟΛΟΓΙΣΜΟΣ')
for g, name, st, mins, _ in finished:
    print(f'  κάρτα {g}  {st:22s} {mins:6.0f} λ   {name}')
print(f'\nσύνολο {(time.time() - t0) / 3600:.1f} h  (εκτίμηση ~{HOURS[STEP]} h)')
for g, name, st, mins, log in finished:
    if st != 'OK':
        print('\n' + '-' * 70)
        print(f'ΑΠΟΤΥΧΙΑ: κάρτα {g} · {name} -- οι τελευταίες γραμμές του {log}:')
        with open(log, encoding='utf-8', errors='replace') as fh:
            print(''.join(fh.readlines()[-25:]))
```
